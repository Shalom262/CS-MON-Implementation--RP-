```python
# Two-head food type and colouring-level prediction
# Food head: Kesari / Muscat / Boondhi
# Colour head: ordered Low < Medium < High

import random
from pathlib import Path
import numpy as np
import pandas as pd
import torch
import torch.nn as nn
from PIL import Image
from sklearn.model_selection import train_test_split
from sklearn.metrics import accuracy_score, f1_score, mean_absolute_error
from torch.utils.data import Dataset, DataLoader
from torchvision import transforms, models

SEED = 42
EPOCHS = 2
BATCH_SIZE = 16
IMAGE_SIZE = 224
DEVICE = torch.device('cuda' if torch.cuda.is_available() else 'cpu')
ROOT = Path.cwd()
CSV_PATH = ROOT / 'details.csv'
IMAGE_DIR = ROOT / 'Dataset'

random.seed(SEED)
np.random.seed(SEED)
torch.manual_seed(SEED)

if not CSV_PATH.exists():
    raise FileNotFoundError(CSV_PATH)

metadata = pd.read_csv(CSV_PATH)
image_column = next(
    column for column in ['Image Name', 'ID', 'Name'] if column in metadata.columns
)
required_columns = [image_column, 'Food', 'Colouring Level']
missing_columns = [column for column in required_columns if column not in metadata]
if missing_columns:
    raise KeyError(f'Missing columns: {missing_columns}')

metadata['image_path'] = metadata[image_column].astype(str).map(
    lambda value: IMAGE_DIR / (value if Path(value).suffix else f'{value}.jpg')
)
metadata = metadata[metadata['image_path'].map(Path.exists)].copy()
metadata['food_id'] = pd.Categorical(metadata['Food']).codes
level_to_id = {'Low': 0, 'Medium': 1, 'High': 2}
id_to_level = {value: key for key, value in level_to_id.items()}
metadata['level_id'] = metadata['Colouring Level'].map(level_to_id)
if metadata['level_id'].isna().any():
    raise ValueError('Colouring Level must contain Low, Medium, or High only.')

food_labels = sorted(metadata['Food'].unique())
metadata['stratify_key'] = metadata['Food'] + '_' + metadata['Colouring Level']
train_df, temporary_df = train_test_split(
    metadata, test_size=0.30, random_state=SEED, stratify=metadata['stratify_key']
)
validation_df, test_df = train_test_split(
    temporary_df, test_size=0.50, random_state=SEED, stratify=temporary_df['stratify_key']
)

train_transform = transforms.Compose([
    transforms.Resize((IMAGE_SIZE, IMAGE_SIZE)),
    transforms.RandomHorizontalFlip(),
    transforms.RandomRotation(8),
    transforms.ColorJitter(brightness=0.15, contrast=0.15),
    transforms.ToTensor(),
    transforms.Normalize([0.485, 0.456, 0.406], [0.229, 0.224, 0.225]),
])
evaluation_transform = transforms.Compose([
    transforms.Resize((IMAGE_SIZE, IMAGE_SIZE)),
    transforms.ToTensor(),
    transforms.Normalize([0.485, 0.456, 0.406], [0.229, 0.224, 0.225]),
])

class SweetDataset(Dataset):
    def __init__(self, frame, transform):
        self.frame = frame.reset_index(drop=True)
        self.transform = transform

    def __len__(self):
        return len(self.frame)

    def __getitem__(self, index):
        row = self.frame.iloc[index]
        image = self.transform(Image.open(row['image_path']).convert('RGB'))
        return {
            'image': image,
            'food': torch.tensor(row['food_id'], dtype=torch.long),
            'level': torch.tensor(row['level_id'], dtype=torch.long),
        }

train_loader = DataLoader(SweetDataset(train_df, train_transform), BATCH_SIZE, shuffle=True)
validation_loader = DataLoader(SweetDataset(validation_df, evaluation_transform), BATCH_SIZE)
test_loader = DataLoader(SweetDataset(test_df, evaluation_transform), BATCH_SIZE)

print(f'Device: {DEVICE}')
print(f'Images: {len(metadata)} | train={len(train_df)} | validation={len(validation_df)} | test={len(test_df)}')
print(f'Food labels: {food_labels}')
```

# CS-MON Experiment Structure

This notebook implements the proposal contribution: **Colour-Space Mamba-Ordinal Network (CS-MON)**.

The system has two prediction heads:

1. **Food type head:** predicts the available sweet categories.
2. **Colouring level head:** predicts the ordered levels `Low < Medium < High` using ordinal learning.

The proposed pipeline is:

`RGB image -> RGB + HSV + LAB -> gated colour attention -> MobileNetV3 feature extraction -> Mamba-inspired feature mixer -> two prediction heads`

The comparison uses a separate RGB ResNet18 baseline. Training is split into separate cells so that each experiment can be explained and evaluated independently.

```python
# 2. Differentiable RGB, HSV, and LAB colour-space conversion
# The input is normalized RGB. It is converted back to [0, 1] before deriving
# the additional colour spaces, so the model never uses a second sensor.
class ColourSpaceConverter(nn.Module):
    def forward(self, normalized_rgb):
        mean = torch.tensor([0.485, 0.456, 0.406], device=normalized_rgb.device).view(1, 3, 1, 1)
        std = torch.tensor([0.229, 0.224, 0.225], device=normalized_rgb.device).view(1, 3, 1, 1)
        rgb = (normalized_rgb * std + mean).clamp(0, 1)
        return torch.cat([rgb, self.rgb_to_hsv(rgb), self.rgb_to_lab(rgb)], dim=1)

    @staticmethod
    def rgb_to_hsv(rgb):
        red, green, blue = rgb.unbind(1)
        maximum = rgb.max(1).values
        minimum = rgb.min(1).values
        delta = (maximum - minimum).clamp_min(1e-6)
        hue = torch.zeros_like(maximum)
        hue = torch.where(maximum == red, ((green - blue) / delta) % 6, hue)
        hue = torch.where(maximum == green, (blue - red) / delta + 2, hue)
        hue = torch.where(maximum == blue, (red - green) / delta + 4, hue)
        hue = (hue / 6) % 1
        saturation = torch.where(maximum > 0, (maximum - minimum) / maximum.clamp_min(1e-6), torch.zeros_like(maximum))
        return torch.stack([hue, saturation, maximum], 1)

    @staticmethod
    def rgb_to_lab(rgb):
        rgb = torch.where(rgb > 0.04045, ((rgb + 0.055) / 1.055).pow(2.4), rgb / 12.92)
        matrix = torch.tensor([[0.4124564, 0.3575761, 0.1804375], [0.2126729, 0.7151522, 0.0721750], [0.0193339, 0.1191920, 0.9503041]], device=rgb.device, dtype=rgb.dtype)
        xyz = torch.einsum('ij,bjhw->bihw', matrix, rgb)
        reference = torch.tensor([0.95047, 1.0, 1.08883], device=rgb.device).view(1, 3, 1, 1)
        ratio = xyz / reference
        threshold = 6 / 29
        transformed = torch.where(ratio > threshold ** 3, ratio.pow(1 / 3), ratio / (3 * threshold ** 2) + 4 / 29)
        lightness = 116 * transformed[:, 1] - 16
        green_red = 500 * (transformed[:, 0] - transformed[:, 1])
        blue_yellow = 200 * (transformed[:, 1] - transformed[:, 2])
        return torch.stack([lightness / 100, (green_red + 128) / 255, (blue_yellow + 128) / 255], 1).clamp(0, 1)
```

```python
# 3. Learnable gated colour-space attention
# A separate gate is learned for each RGB, HSV, and LAB channel. This makes
# the model inspect which colour evidence is useful under each condition.
class GatedColourAttention(nn.Module):
    def __init__(self, channels=9):
        super().__init__()
        self.pool = nn.AdaptiveAvgPool2d(1)
        self.gate = nn.Sequential(nn.Conv2d(channels, channels, 1), nn.Sigmoid())

    def forward(self, colour_spaces):
        weights = self.gate(self.pool(colour_spaces))
        return colour_spaces * weights, weights.flatten(1)

print('Colour groups: RGB(3) + HSV(3) + LAB(3) = 9 channels')
```

```python
# 4. Mamba-inspired state-space feature mixer
# This lightweight depthwise sequence mixer operates on CNN feature tokens.
# It is Mamba-inspired, not a claim that the full external Mamba package is used.
class MambaInspiredMixer(nn.Module):
    def __init__(self, feature_count):
        super().__init__()
        self.norm = nn.LayerNorm(feature_count)
        self.input_projection = nn.Linear(feature_count, feature_count * 2)
        self.state_projection = nn.Conv1d(feature_count, feature_count, 3, padding=1, groups=feature_count)
        self.output_projection = nn.Linear(feature_count, feature_count)

    def forward(self, feature_map):
        batch, channels, height, width = feature_map.shape
        tokens = feature_map.flatten(2).transpose(1, 2)
        content, gate = self.input_projection(self.norm(tokens)).chunk(2, dim=-1)
        state = self.state_projection(content.transpose(1, 2)).transpose(1, 2)
        mixed = self.output_projection(torch.tanh(state) * torch.sigmoid(gate))
        return (tokens + mixed).transpose(1, 2).reshape(batch, channels, height, width)
```

```python
# 5A. Pretrained-weight helper
# If weights are unavailable offline, torchvision falls back to random initialization.
def pretrained_or_none(weight_class):
    try:
        return weight_class.DEFAULT
    except Exception:
        return None
```

```python
# 5. CS-MON proposal model and RGB baseline
class OrdinalHead(nn.Module):
    def __init__(self, feature_count):
        super().__init__()
        self.output = nn.Linear(feature_count, 2)

    def forward(self, features):
        return self.output(features)

class TwoHeadModel(nn.Module):
    def __init__(self, backbone, feature_count):
        super().__init__()
        self.backbone = backbone
        self.food_head = nn.Linear(feature_count, len(food_labels))
        self.colour_level_head = OrdinalHead(feature_count)

    def forward(self, images):
        features = self.backbone(images)
        if features.ndim > 2:
            features = torch.flatten(features, 1)
        return {'food': self.food_head(features), 'colour_level': self.colour_level_head(features)}

class CS_MON(nn.Module):
    def __init__(self, food_count):
        super().__init__()
        self.colour_spaces = ColourSpaceConverter()
        self.gated_attention = GatedColourAttention(9)
        self.input_projection = nn.Sequential(nn.Conv2d(9, 3, 1), nn.BatchNorm2d(3), nn.ReLU(inplace=True))
        backbone = models.mobilenet_v3_small(weights=pretrained_or_none(models.MobileNet_V3_Small_Weights))
        self.feature_extractor = backbone.features
        self.mamba_mixer = MambaInspiredMixer(576)
        self.pool = nn.AdaptiveAvgPool2d(1)
        self.food_head = nn.Linear(576, food_count)
        self.colour_level_head = OrdinalHead(576)

    def forward(self, images):
        colour_spaces = self.colour_spaces(images)
        gated, gate_weights = self.gated_attention(colour_spaces)
        features = self.feature_extractor(self.input_projection(gated))
        features = self.mamba_mixer(features)
        features = self.pool(features).flatten(1)
        return {'food': self.food_head(features), 'colour_level': self.colour_level_head(features), 'gate_weights': gate_weights}

def build_rgb_resnet_baseline():
    backbone = models.resnet18(weights=pretrained_or_none(models.ResNet18_Weights))
    feature_count = backbone.fc.in_features
    backbone.fc = nn.Identity()
    return TwoHeadModel(backbone, feature_count).to(DEVICE)

def build_cs_mon():
    return CS_MON(len(food_labels)).to(DEVICE)

print('The baseline uses RGB-ResNet18.')
print('The proposal uses RGB-HSV-LAB + gated attention + MobileNetV3 + Mamba-inspired mixing.')
```

```python
# 6. Loss functions, ordinal encoding, and evaluation metrics
# CORAL-style ordinal targets encode Low as [0, 0], Medium as [1, 0],
# and High as [1, 1]. This preserves the natural distance between levels.
def pretrained_or_none(weight_class):
    try:
        return weight_class.DEFAULT
    except Exception:
        return None

def ordinal_targets(level_ids):
    return torch.stack([level_ids >= 1, level_ids >= 2], dim=1).float()

def decode_levels(logits):
    return (torch.sigmoid(logits) >= 0.5).sum(dim=1).long()

food_loss = nn.CrossEntropyLoss()
colour_loss = nn.BCEWithLogitsLoss()

def calculate_metrics(food_true, food_pred, level_true, level_pred):
    return {
        'food_accuracy': accuracy_score(food_true, food_pred),
        'food_macro_f1': f1_score(food_true, food_pred, average='macro', zero_division=0),
        'colour_level_accuracy': accuracy_score(level_true, level_pred),
        'colour_level_mae': mean_absolute_error(level_true, level_pred),
        'joint_accuracy': np.mean((np.asarray(food_true) == np.asarray(food_pred)) & (np.asarray(level_true) == np.asarray(level_pred))),
    }
```

```python
# 7. One-epoch runner and early-stopping training utility
# Early stopping monitors validation joint accuracy. The best checkpoint is
# restored, preventing over-training and making the reported test result fair.
def run_epoch(model, loader, optimizer=None):
    training = optimizer is not None
    model.train(training)
    total_loss = 0.0
    food_true, food_pred, level_true, level_pred = [], [], [], []

    for batch in loader:
        images = batch['image'].to(DEVICE)
        food = batch['food'].to(DEVICE)
        level = batch['level'].to(DEVICE)
        with torch.set_grad_enabled(training):
            output = model(images)
            loss = food_loss(output['food'], food) + colour_loss(output['colour_level'], ordinal_targets(level))
            if training:
                optimizer.zero_grad()
                loss.backward()
                torch.nn.utils.clip_grad_norm_(model.parameters(), 1.0)
                optimizer.step()
        total_loss += loss.item() * len(images)
        food_true.extend(food.cpu().numpy())
        food_pred.extend(output['food'].argmax(1).detach().cpu().numpy())
        level_true.extend(level.cpu().numpy())
        level_pred.extend(decode_levels(output['colour_level']).detach().cpu().numpy())

    metrics = calculate_metrics(food_true, food_pred, level_true, level_pred)
    metrics['loss'] = total_loss / len(loader.dataset)
    return metrics

def train_with_early_stopping(model, model_name, epochs=20, patience=4):
    optimizer = torch.optim.AdamW(model.parameters(), lr=3e-4, weight_decay=1e-4)
    scheduler = torch.optim.lr_scheduler.ReduceLROnPlateau(optimizer, mode='max', patience=1, factor=0.5)
    best_score = -np.inf
    best_state = None
    wait = 0
    history = []

    for epoch in range(1, epochs + 1):
        train_metrics = run_epoch(model, train_loader, optimizer)
        validation_metrics = run_epoch(model, validation_loader)
        score = validation_metrics['joint_accuracy']
        scheduler.step(score)
        history.append({'epoch': epoch, **{f'train_{k}': v for k, v in train_metrics.items()}, **{f'val_{k}': v for k, v in validation_metrics.items()}})
        print(f"{model_name} | epoch {epoch:02d} | train joint={train_metrics['joint_accuracy']:.3f} | val joint={score:.3f}")

        if score > best_score:
            best_score = score
            best_state = {key: value.detach().cpu().clone() for key, value in model.state_dict().items()}
            wait = 0
        else:
            wait += 1
            if wait >= patience:
                print(f'Early stopping: no validation improvement for {patience} epochs.')
                break

    model.load_state_dict(best_state)
    return model, pd.DataFrame(history), run_epoch(model, test_loader)
```

```python
# 8. Train the RGB ResNet18 baseline separately
RUN_BASELINE = True
BASELINE_EPOCHS = 20
BASELINE_PATIENCE = 4

baseline_model = None
baseline_history = pd.DataFrame()
baseline_test_metrics = {}

if RUN_BASELINE:
    baseline_model = build_rgb_resnet_baseline()
    baseline_model, baseline_history, baseline_test_metrics = train_with_early_stopping(
        baseline_model, 'RGB-ResNet18 baseline', BASELINE_EPOCHS, BASELINE_PATIENCE
    )
    print('Baseline test metrics:', baseline_test_metrics)
else:
    print('Baseline training skipped.')
```

```python
# 9. Train the proposed CS-MON model separately
RUN_CS_MON = True
CS_MON_EPOCHS = 20
CS_MON_PATIENCE = 4

cs_mon_model = None
cs_mon_history = pd.DataFrame()
cs_mon_test_metrics = {}

if RUN_CS_MON:
    cs_mon_model = build_cs_mon()
    cs_mon_model, cs_mon_history, cs_mon_test_metrics = train_with_early_stopping(
        cs_mon_model, 'CS-MON RGB-HSV-LAB-Mamba', CS_MON_EPOCHS, CS_MON_PATIENCE
    )
    print('CS-MON test metrics:', cs_mon_test_metrics)
else:
    print('CS-MON training skipped.')
```

```python
# 10. Compare the baseline and CS-MON test performance
comparison_rows = []
if baseline_model is not None:
    comparison_rows.append({'model': 'RGB-ResNet18 baseline', **baseline_test_metrics})
if cs_mon_model is not None:
    comparison_rows.append({'model': 'CS-MON RGB-HSV-LAB-Mamba', **cs_mon_test_metrics})
comparison_table = pd.DataFrame(comparison_rows)
display(comparison_table)

# The joint score requires both heads to be correct for the same image.
print('Evaluation metrics: food accuracy, food macro F1, colouring-level accuracy, ordinal MAE, and joint accuracy.')
```

```python
# 13. Final two-head inference output
selected_model = cs_mon_model if cs_mon_model is not None else baseline_model
selected_model_name = 'CS-MON' if cs_mon_model is not None else 'RGB baseline'

if selected_model is not None:
    selected_model.eval()
    sample = test_df.iloc[0]
    image = Image.open(sample['image_path']).convert('RGB')
    input_tensor = evaluation_transform(image).unsqueeze(0).to(DEVICE)
    with torch.no_grad():
        output = selected_model(input_tensor)
        food_probability = torch.softmax(output['food'], dim=1)[0]
        ordinal_probability = torch.sigmoid(output['colour_level'])[0]

    predicted_food_id = int(food_probability.argmax())
    predicted_level_id = int((ordinal_probability >= 0.5).sum())
    result = {
        'model': selected_model_name,
        'image': str(sample['image_path']),
        'food_prediction': food_labels[predicted_food_id],
        'food_confidence': round(float(food_probability[predicted_food_id]), 4),
        'colouring_level_prediction': id_to_level[predicted_level_id],
        'ordinal_threshold_probabilities': ordinal_probability.cpu().numpy().round(4).tolist(),
        'screening_notice': 'Image-based screening only; laboratory analysis is required for chemical confirmation.',
    }
    print(result)
else:
    print('Train at least one model before inference.')
```
